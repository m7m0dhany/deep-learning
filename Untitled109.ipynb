```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

class SimpleLSTM:
    def init(self):
        self.wf_x = 0.5
        self.wf_h = 0.1
        self.bf = 0
        
        self.wi_x = 0.6
        self.wi_h = 0.2
        self.bi = 0
        
        self.wc_x = 0.7
        self.wc_h = 0.3
        self.bc = 0
        
        self.wo_x = 0.8
        self.wo_h = 0.4
        self.bo = 0
        
        self.wy = 0.5
        self.by = 0
        
        self.h = 0
        self.c = 0
        
    def forward_step(self, x):
        f_gate = sigmoid(self.wf_x * x + self.wf_h * self.h + self.bf)
        i_gate = sigmoid(self.wi_x * x + self.wi_h * self.h + self.bi)
        c_candidate = tanh(self.wc_x * x + self.wc_h * self.h + self.bc)
        self.c = f_gate * self.c + i_gate * c_candidate
        o_gate = sigmoid(self.wo_x * x + self.wo_h * self.h + self.bo)
        self.h = o_gate * tanh(self.c)
        
        return {
            'f_gate': f_gate,
            'i_gate': i_gate,
            'c_candidate': c_candidate,
            'c_state': self.c,
            'o_gate': o_gate,
            'h_state': self.h
        }
    
    def predict(self, h):
        return self.wy * h + self.by

lstm = SimpleLSTM()
X = [1, 2, 3]

for t, x in enumerate(X, 1):
    gates = lstm.forward_step(x)
    print(f"\nTime Step t={t}, Input x{t}={x}")
    print(f"f_{t} = {gates['f_gate']:.4f}")
    print(f"i_{t} = {gates['i_gate']:.4f}")
    print(f"C̃_{t} = {gates['c_candidate']:.4f}")
    print(f"C_{t} = {gates['c_state']:.4f}")
    print(f"o_{t} = {gates['o_gate']:.4f}")
    print(f"h_{t} = {gates['h_state']:.4f}")

gates_pred = lstm.forward_step(4)
print(f"\nTime Step t=4, Input x4=4")
print(f"f_4 = {gates_pred['f_gate']:.4f}")
print(f"i_4 = {gates_pred['i_gate']:.4f}")
print(f"C̃_4 = {gates_pred['c_candidate']:.4f}")
print(f"C_4 = {gates_pred['c_state']:.4f}")
print(f"o_4 = {gates_pred['o_gate']:.4f}")
print(f"h_4 = {gates_pred['h_state']:.4f}")

predicted_value = lstm.predict(lstm.h)
print(f"\nPrediction for x4 = {predicted_value:.4f}")
```

```python
import numpy as np
def sigmoid(x):
return 1 / (1 + np.exp(-x))
def tanh(x):
return np.tanh(x)
Wf = 0.5; Whf = 0.1; bf = 0
Wi = 0.6; Whi = 0.2; bi = 0
Wc = 0.7; Whc = 0.3; bc = 0
Wo = 0.8; Who = 0.4; bo = 0
Wy = 4; by = 0
X = [1, 2, 3, 4] 
h_prev = 0 # h_{t-1}
C_prev = 0 # C_{t-1}
print("--- Manual LSTM Implementation ---")
for t, x_t in enumerate(X):
print(f"\nTime Step t={t+1}, Input x_t={x_t}")
f_t = sigmoid(Wf * x_t + Whf * h_prev + bf)
print(f"Forget Gate (f_t): {f_t:.3f}")
i_t = sigmoid(Wi * x_t + Whi * h_prev + bi)
print(f"Input Gate (i_t): {i_t:.3f}")
C_tilde_t = tanh(Wc * x_t + Whc * h_prev + bc)
print(f"Candidate Cell State (C_tilde_t): {C_tilde_t:.3f}")
C_t = f_t * C_prev + i_t * C_tilde_t
print(f"Cell State (C_t): {C_t:.3f}")
o_t = sigmoid(Wo * x_t + Who * h_prev + bo)
print(f"Output Gate (o_t): {o_t:.3f}")
h_t = o_t * tanh(C_t)
print(f"Hidden State (h_t): {h_t:.3f}")
# Update for next time step
h_prev = h_t
C_prev = C_t
y_hat = Wy * h_prev + by
print(f"\n--- Final Prediction ---")
print(f"Predicted next value (y_hat): {y_hat:.3f}")
```
