```python
import numpy as np

def softmax(x):
    x = x - np.max(x, axis=1, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)
```

```python
def compute_attention(sentence):
    tokens = sentence.split()
    np.random.seed(42)
    X = np.random.rand(len(tokens), 4)
    d_k = X.shape[1]
    W_Q = np.random.rand(d_k, d_k)
    W_K = np.random.rand(d_k, d_k)
    W_V = np.random.rand(d_k, d_k)
    Q = np.dot(X, W_Q)
    K = np.dot(X, W_K)
    V = np.dot(X, W_V)
    scores = np.dot(Q, K.T) / np.sqrt(d_k)
    weights = softmax(scores)
    output = np.dot(weights, V)
    return tokens, scores, weights, output
```

```python
def run():
    sentence = input("Enter sentence: ")
    if sentence.strip() == "":
        print("Empty input!")
        return
    tokens, scores, weights, output = compute_attention(sentence)
    print("\nTOKENS:")
    print(tokens)
    print("\nATTENTION SCORES:")
    print(scores)
    print("\nATTENTION WEIGHTS:")
    print(weights)
    print("\nOUTPUT (CONTEXT VECTORS):")
    print(output)

if __name__ == "__main__":
    run()
```
